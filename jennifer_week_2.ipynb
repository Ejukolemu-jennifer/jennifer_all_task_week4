```python
def calculator_argument(numbera,numberb:int):
    operation=input("enter an operation (+,-,*,/)")
    if operation =="+":
        print(f'result:{numbera+numberb}')
    elif operation =="-":
        print(f'result{numbera-numberb}')
    elif operation =="*":
        print(f'result{numbera*numberb}')
    elif operation =='/':
        if numberb==0:
            print("cannot divide by zero")
        else:
            print(f'result{numbera / numberb}')
       
    
calculator_argument(20,5)
```

```python
def tax_checker():
    monthly_salary=int(input("enter your salary"))
    if monthly_salary> 250000:
        tax=monthly_salary * 0.15
        net_salary = monthly_salary - tax
        return net_salary
    else:
        tax = monthly_salary*0.05
        net_salary=monthly_salary - tax
        return net_salary
    
tax_checker()
```
