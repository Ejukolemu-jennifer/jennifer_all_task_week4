```python
class Rectangle:
    def __init__(self,width,height):
        self.width=width
        self.height=height

    def area(self):
        return f'{self.width*self.height}'
    
    def parameter(self):
        return f'{2*(self.width+self.height)}'


shape=Rectangle(4,2)
print(shape.area())

shape=Rectangle(4,2)
print(shape.parameter())




```
